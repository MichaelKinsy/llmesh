```python
import nest_asyncio # Enables async in Jupyter
nest_asyncio.apply()

import httpx

class SimpleClient:
    def __init__(self, base_url="http://localhost:8000"):
        self.base_url = base_url
        
    async def download_model(self, repo_id):
        async with httpx.AsyncClient(timeout=300) as client:
            response = await client.post(f"{self.base_url}/models/download", params={"repo_id": repo_id})
            return response.json()
    
    async def load_model(self, model_name):
        async with httpx.AsyncClient(timeout=300) as client:
            response = await client.post(f"{self.base_url}/models/{model_name}/load")
            return response.json()
    
    async def generate(self, model_name, prompt, max_tokens=100):
        async with httpx.AsyncClient(timeout=300) as client:
            payload = {"name": model_name, "prompt": prompt, "max_tokens": max_tokens}
            response = await client.post(f"{self.base_url}/generate", json=payload)
            return response.text
    
    async def chat(self, model_name, messages, max_tokens=100):
        async with httpx.AsyncClient(timeout=300) as client:
            payload = {"name": model_name, "messages": messages, "max_tokens": max_tokens}
            response = await client.post(f"{self.base_url}/chat", json=payload)
            return response.text

client = SimpleClient()
print("✅ Client ready")
```

```python
# Download BitNet model
print("📥 Downloading model...")
result = await client.download_model("microsoft/BitNet-b1.58-2B-4T-gguf")
print(f"Download result: {result}")

# Load the model
print("\n🔄 Loading model...")
model_name = "BitNet-b1.58-2B-4T"
load_result = await client.load_model(model_name)
print(f"Load result: {load_result}")

prompt = "System: You are a pirate impersonator | Question: How would a stereotypical pirate describe the ocean? | Answer:"
print(f"Prompt: {prompt}")
print("Response:")

response = await client.generate(model_name, prompt, max_tokens=150)
print(response)
```

```python
print("📥 Downloading TinyLlama model...")
result = await client.download_model("TheBloke/TinyLlama-1.1B-Chat-v1.0-GGUF")
print(f"Download result: {result}")

print("\n🔄 Loading model...")
model_name = "TinyLlama-1.1B-Chat-v1.0-GGUF"  # Note: keeps the -GGUF suffix for non-BitNet models
load_result = await client.load_model(model_name)
print(f"Load result: {load_result}")

response = await client.generate(
    model_name=model_name,
    prompt="System: You are a pirate impersonator | Question: How would a stereotypical pirate describe the ocean? | Answer:",
    max_tokens=150
)
print("Generated text:", response)
```
